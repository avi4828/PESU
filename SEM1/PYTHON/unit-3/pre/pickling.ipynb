import the module

```python
import pickle as p
```

method used 1. dump 2.dumps 3.load 4.loads

1.dump - serialize the data into a file 

2. dumps - serialize the data in memory in secondry memory

3.load - deserialize the data from a file

4.loads = deserialize the data from memory

1. numeric data

```python
a=10;b=23.5;c=9+8j
p_a=p.dumps(a)
p_b=p.dumps(b)
p_c=p.dumps(c)
type(a),type(b),type(c)
```

```python
type(p_a),type(p_b),type(p_c)
```

```python
print(p_a)
print(p_b)
print(p_c)
```

```python
import sys
```

```python
print(sys.getsizeof(a),sys.getsizeof(p_a))
print(sys.getsizeof(b),sys.getsizeof(p_b))
print(sys.getsizeof(c),sys.getsizeof(p_c))
```

```python
d_a=p.loads(p_a)
d_b=p.loads(p_b)
d_c=p.loads(p_c)
print(d_a,d_b,d_c)
type(d_a),type(d_b),type(d_c)
```

2. sequence data

```python
l=[1,2,3,'banana',[23,45.7,9-7j]]
p_l=p.dumps(l)
print(type(l),type(p_l))
d_l=p.loads(p_l)
print(l,p_l,d_l,type(d_l))
```

```python
d = {1:'a',2:2,3:52.2}
p_d = p.dumps(d)
print(type(d),type(p_d))
```

```python
print(p_d)
```

```python
d_d = p.loads(p_d)
print(d_d)
```

picking the data into a file

```python
l1=[1,2,3,'apple']
d1={1:'a',2:'b'}
s1='Hello picking the data'
n1=56.89898
with open('data.pkl','wb')as f:
    p.dump(l1,f)
    p.dump(d1,f)
    p.dump(s1,f)
    p.dump(n1,f)
```

```python
with open('data.pkl','rb')as f:
    d_l1=p.load(f)
    d_d1=p.load(f)
    d_s1=p.load(f)
    d_n1=p.load(f)
    print(d_l1,d_d1,d_s1,d_n1)
```

4. custom object


```python
class person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def __repr__(self):
        return f'Person(name:{self.name},age:{self.age})'

p1=person('alpha',21)
```

```python
type(p1)
```

```python

p_p1=p.dumps(p1)
type(p_p1),p_p1
```
