Writelines()


```python
# writelines use to enter multiple line in file 
# witrelines take 1 argument - is lines or string that was store in list or tupple

lines=[
    'frist lines\n',
    'second lines\n',
    'thrid lines\n'
]
with open('file1.txt','w') as f:
    f.writelines(lines)
f=open('file1.txt','r')
str = f.read()
print(str)
f.close()
```
