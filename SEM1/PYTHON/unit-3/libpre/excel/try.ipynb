openpyxl is use to read and write a excel file openpyxl is use to read and write .xlsx thay can not read or write a .xls file


```python
import openpyxl as op
```

```python
wb = op.load_workbook('VideoSales1.xlsx')
```

```python
ws = wb.active
print(ws,type(ws))
```

```python
ws = wb['SalesData']
```

```python
print(ws.max_column)
print(ws.max_row)
```

```python
print(f'the D5 cell data values is {ws['b5'].value}')
```

```python
data = [ws.cell(row=i,column=2).value for i in range(1,ws.max_row+1)]
for d in data : print(d)
```

```python
rpos = 2
cpos = 7

t_sales = ((ws.cell(row=rpos,column=cpos).value)+
            (ws.cell(row=rpos,column=cpos+1).value)+
            (ws.cell(row=rpos,column=cpos+2).value)+
            (ws.cell(row=rpos,column=cpos+3).value)
            )
print(t_sales)
ws.cell(row=rpos,column=cpos+4).value = t_sales
wb.save('VideoSales1.xlsx')
```
