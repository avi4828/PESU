```python
import gspread
from google.oauth2.service_account import Credentials
from collections import Counter
```

use the downloaded key file

```python
SERVICE_ACCOUNT_FILE = 'services_account.json'
```

use the google sheet url

```python
SHEET_URL = 'https://docs.google.com/spreadsheets/d/1KZCb8IqiyN0MGOStKAiMHX79MGahmCziew7rQRGo21k/edit?usp=sharing'
```

identify the sheet tab to be read

```python
WORKSHEET_NAME = 'Sheet1'
```

Define read only access to Sheets and drive

```python
SCOPES = [
    "https://www.googleapis.com/auth/spreadsheets.readonly",
    "https://www.goolgeapis.com/auth/drive.readonly",
]
```

```python
def get_preferences():
    creds = Credentials.from_service_account_file(SERVICE_ACCOUNT_FILE,scopes=SCOPES)
    # credential are lded form the json file oauth scopes are applied
    client = gspread.authorize(creds)
    # gspread client is authorized
    spreadsheet = client.open_by_url(SHEET_URL)
    # open the sheet using the url
    ws = spreadsheet.worksheet(WORKSHEET_NAME)
    # select the sheet by its name
    rows = ws.get_all_values()
    # read the entire sheet as a list of rows
    if not rows:
        raise RuntimeError('not data found')
    header = rows[0]
    # first row is treated as header row 
    data_rows = rows[1:]
    # remaining rows are treated as data
    if 'ChannelPreference' not in header:
        raise RuntimeError('column not found')
    index = header.index('ChannelPreference')
    preference=[]
    for r in data_rows:
        if len(r)>index:
            val=r[index].strip()
            if val != "":
                preference.append(val)
    c=Counter(preference)
    print('ChannelPreference values')
    for p in preference:
        print(p)
    for k,v in c.most_common():
        print(f'{k},{v}')
        return preference,c
    
```

```python
get_preferences()
```
