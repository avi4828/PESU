3.xml file

```python
import xml.etree.ElementTree as ET
```

```python
data = response.json()
```

```python
root = ET.Element('Posts')
for item in data['posts']:
    pel = ET.SubElement(root,'Post')
    iel = ET.SubElement(pel,'ID')
    iel.text = str(item['id'])
    uet = ET.SubElement(pel,'User ID')
    uet.text = str(item['userId'])
    tel =  ET.SubElement(root,'Title')
    tel.text = str(item['title'])
    vel = ET.SubElement(root,'Views')
    vel.text = str(item['views'])
tree = ET.ElementTree(root)
tree.write('api_1_data.xml',encoding='utf-8',xml_declaration=False)
```

managing API Limits


```python
import time
```

```python
max = 3
dalay = 2 # seconds
for attempt in range(max):
    response = req.get("http://dummyjson.com/posts/2")
    print(response.status_code)
    if response.status_code == 429:
        print(f'rate Limited waiting for {dalay} seconds')
        time.sleep(dalay)
        dalay*=2
    else:
        break
```
